```python
from vllm import LLM
from cs336_alignment.math_baseline import format_prompt, generation_hyperparameters, evaluate_vllm
from cs336_alignment.drgrpo_grader import r1_zero_reward_fn

math_path = "/home/ansonl32/ECE405/ece405-assignment3-alignment/data/math/math.parquet"
prompt_file = "/home/ansonl32/ECE405/ece405-assignment3-alignment/cs336_alignment/prompts/r1_zero.prompt"
model_path = "/home/ansonl32/koa_scratch/ECE405/assignment3/Qwen/Qwen2.5-Math-1.5B"
output_file = "/home/ansonl32/ECE405/ece405-assignment3-alignment/cs336_alignment/outputs/zero_shot_baseline_results.jsonl"

# Obtains both the data from MATH and the formatted prompts
examples, formatted_prompts = format_prompt(math_path, prompt_file)
sampling_params = generation_hyperparameters()


print(f"Loading model {model_path}...")
llm = LLM(model=model_path)


evaluate_vllm(
    vllm_model=llm,
    reward_fn=r1_zero_reward_fn,
    examples=examples,
    prompts=formatted_prompts,
    eval_sampling_params=sampling_params,
    output_path=output_file
)
```
